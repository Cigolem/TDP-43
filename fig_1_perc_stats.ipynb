```python
# STEP 1 — Load data

import os
import pandas as pd
import statsmodels.formula.api as smf
import numpy as np

os.chdir(r"C:/Users/cigde/Dropbox/PDC_Projects/Jemeen/sleep_python/")
df_raw = pd.read_csv("sleep_scores.csv")


# STEP 2: Keep only needed columns
df = df_raw[['Mouse_ID', 'Group', 'Phase', 'Hour', 'Hour24',
             'Wake_perc', 'NREM_perc', 'REM_perc']].copy()

# Make sure data types are correct
df['Hour24'] = df['Hour24'].astype(int)
df['Group'] = df['Group'].astype('category')
df['Mouse_ID'] = df['Mouse_ID'].astype('category')

# STEP 3 — Check for missing data and group distributoion
print("Missing values per column:")
print(df.isnull().sum())
print(f"\nTotal observations: {len(df)}")
print(f"Number of mice: {df['Mouse_ID'].nunique()}")
print("\nMice per group:")
print(df.groupby('Group', observed=True)['Mouse_ID'].nunique())
```

```python
# STEP 2 — Use repeated measures ANOVA approach - WAKE

from scipy import stats
import numpy as np

print("\n" + "="*80)
print("Wake PERCENTAGE - Repeated Measures Analysis")
print("="*80)

# Reshape data for repeated measures: one row per mouse
df_wide = df.pivot_table(
    values='Wake_perc',
    index=['Mouse_ID', 'Group'],
    observed=True,
    columns='Hour24'
).reset_index()

print(f"\nData shape: {df_wide.shape}")
print(f"WT mice: {(df_wide['Group'] == 'WT').sum()}")
print(f"Q331K mice: {(df_wide['Group'] == 'HOM').sum()}")

# Separate by group
wt_data = df_wide[df_wide['Group'] == 'WT'].iloc[:, 2:].values
q331k_data = df_wide[df_wide['Group'] == 'HOM'].iloc[:, 2:].values

# Two-way repeated measures ANOVA using pingouin
import pingouin as pg

aov_wake = pg.mixed_anova(
    data=df,
    dv='Wake_perc',
    within='Hour24',
    between='Group',
    subject='Mouse_ID'
)

print("\n" + "="*60)
print("TWO-WAY REPEATED MEASURES ANOVA RESULTS - Wake")
print("="*60)

# Display with all columns (to see epsilon and GG-corrected p-values)
print("\nWAKE ANOVA table ")
print(aov_wake.to_string())
```
